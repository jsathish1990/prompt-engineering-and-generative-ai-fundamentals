#### **Problem Statement:**

- Prompt Engineering is the technique of structuring text inputs - that can be interpreted and understood by generative AI models.

- Prompt Engineering involves developing and optimizing prompts to efficiently utliize large language models (LLM) for various applications.

- Some examples of Large Language Models are : Gemini, BERT, GPT-3, Llama, etc.

- Large refers to the number of parameters in the model. For example : BERT has 110M parameters, GPT-3 has 175B parameters and so on.

#### **Best Practices in Prompt Engineering:**

- Being as specific and clear as possible.
- Structure the prompt in a way to provide the context around the question.
- Use specific examples to help the model provide accurate responses.
- Use of one-shot or few-shot prompting techniques.
- Iterate and refine based on the response received from the large language model.

#### **Prompt Techniques**

- **Zero-shot** prompting technique : The large language model provides a response based on  given instructions and context but no examples provided.

- Some nuanced tasks require examples as additional context for the model to improve the performance and providing only instructions to the model may not suffice. This type of prompting technique is called **Few-Shot** prompting.

- Although providing example to the large language model could help understand the patterns, few-shot prompting may not suffice for complex reasoning tasks.

- **Chain-of-thought** prompting technique : This prompting technique leads a model to additional reasoning steps, improving the response from the model.

- This can be achieved by detailed examples on how to approach the problem with additional phrases such as "work through the problem step-by-step".


- **Tree-of-thoughts** prompting technique : For complex reasoning tasks where traditional simpler apporaches may not be sufficient, tree-of-thoughts approach has been recently proposed in literature, encouraging trial and error and backtracking when necessary.

- **Retrieval Augmented Generation  (RAG)** : For complex knowledge intensive tasks, an advanced prompt engineering technique called RAG uses external knowledge sources for factual consistency, improving the response from the large language model and mitigating the problem of hallucination.

#### **Chain of Thought Prompting**

- Enables complex reasoning through intermediate reasoning steps.

- Providing LLM with some chain-of-thoughts examples as part of the prompt.

- **Wei, Jason, et al. "Chain-of-thought prompting elicits reasoning in large language models." Advances in Neural Information Processing Systems 35 (2022): 24824-24837.**


```python
# Chain of thought prompting where example is provided to show the model to think through step-by-step.

prompt = """ Your task is to think through the problem step by step. Take a
             systematic approach.

             Question : A shopping cart has 5 pencils, 10 notebooks, 3 blue
             pens, 4 ink bottles and 6 sticky notes. How many total items are
             in the cart?

             Response : The shopping cart has 5 pencils + 10 notebooks + 3
             blue pens + 4 ink bottles + 6 sticky notes, which add up to 28
             items.

             Question : A shelf has 20 notebooks. If 3 notebooks were taken by
             someone and three times the remaining were added to the shelf, how
             many notebooks remain on the shelf?

             Response :   """
```

```python
%%time
# Extracting model response using model.generate_content() method

response = model.generate_content(prompt)
```

```python
# Display the model response
to_markdown(response.text)
```
