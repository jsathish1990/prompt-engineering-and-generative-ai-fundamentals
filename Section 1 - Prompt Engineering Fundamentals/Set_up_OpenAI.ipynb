#### **Problem Statement:**

- Prompt Engineering is the technique of structuring text inputs - which can be interpreted and understood by a generative AI model.

- Prompt Engineering involves developing and optimizing prompts to efficiently use large language models (LLM) for varied applications.

- Some examples of Large Language Models are : Gemini, Bert, GPT-3, Llama, etc.

- "Large" refers to the number of parameters in the model. For example : Bert has 110M parameters, GPT-3 has 175B parameters and so on.

#### **Best Practices in Prompt Engineering:**

- Being as specific and clear as possible.
- Structuring the prompt in a way to provide the context around the question.
- Using specific examples to help the model provide accurate responses.
- Use of one-shot or few-shot prompting techniques.
- Iterate and refine based on the response received from the large language model.


- https://www.promptingguide.ai/


- Ekin, Sabit. "Prompt engineering for ChatGPT: A quick guide to techniques, tips, and best practices." Authorea Preprints (2023).

#### **OpenAI API KEY**

```python
# https://platform.openai.com/api-keys
# Setting up the OpenAI API KEY, saved in Google Colab notebook

# Extract OpenAI API Key
OPENAI_API_KEY=userdata.get('OPENAI_API_KEY')

# Set the OpenAI library to use the API Key and initializing the OpenAI Client/
client = OpenAI(
    api_key = OPENAI_API_KEY
)
```

```python
# client.chat.completions.create() method where the model name is specified
# messages has a list of dictionaries consisting of system and user prompts.

chat_completion = client.chat.completions.create(
    model="gpt-4o-mini", #specify the model name
    messages=[
        {"role": "system","content":"You are a helpful assistant"},
        {"role": "user", "content": "What is artificial intelligence?"},
        {"role": "system", "content": "Artificial intelligence is the simulation of human intelligence processes by computer systems"},
        {"role": "user", "content": "What are two applications of artificial intelligence?"}
]
)
```

```python
# Response from the model extracted using `chat_completions.choices` accessor

html_content = chat_completion.choices[0].message.content
```

```python
# Display the response using `display()` and `HTML()` methods from IPython.display package

display(HTML(html_content))
```

#### **Tree-of-thoughts prompting**

- For complex reasoning tasks where traditional simpler approaches may not be sufficient, tree-of-throughts prompting technique has been recently proposed in literature, which allows to backtrack when necessary encouraging trial and error.

-  **Long, Jieyi. "Large Language Model Guided Tree-of-Thought." arXiv preprint arXiv:2305.08291 (2023).**

- **Yao, Shunyu, et al. "Tree of thoughts: Deliberate problem solving with large language models." arXiv preprint arXiv:2305.10601 (2023).**

- https://www.promptingguide.ai/techniques/tot

#### **Basic Prompt**

```python
# Tree of thoughts - prompting technique

prompt = """
         Imagine three different experts are thinking about this question:

         All experts will write down one step of their thought process,

         then share it with the group.

         Then the experts will proceed to the next step.

         If at any point, an expert realizes they are incorrect, then they leave.

         The question is :

         1) Alex is sitting in the office desk
         2) Alex takes a key from the drawer and walks to the storage room.
         3) In the storage room, Alex unlocks a box and puts a notebook inside.
         4) Alex locks the box and returns to the office desk without the key.
         5) Then, Alex goes to have food at the cafeteria
         6) Where is the key?
         """
```

```python
# Openai text genertion model : client.chat.completions.create()

chat_completions = client.chat.completions.create(
    model = "gpt-4o-mini",
    messages = [
        {
            "role": "user",
            "content": prompt
        }
    ]
)
```

```python
# Response from the model
html_content = chat_completions.choices[0].message.content
```

```python
# Display the response using `display()` and `HTML()` methods from IPython.display() package

display(HTML(html_content))
```
