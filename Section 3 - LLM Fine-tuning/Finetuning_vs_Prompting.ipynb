#### **Problem Statement**

- Large language models such as GPT, Llama are examples of pre-trained transformer models, trained to predict the next word or token, given some input text.

- These models typically have millions or billions of parameters and are quite versatile, useful for various tasks in NLP by instructing these models through appropriate natural language prompts.

- Prompt Engineering is an iterative - process, requiring clear instructions and being as detailed as possible.

- Even some modifications to prompt could yield different outputs.

- Llama, Llama2, GPT-2 are some examples of decoder-only transformer based models, where as BART, Flan-T5 are examples of encoder-decoder type transformer based models, useful for tasks where the model output heavily depends on the input (Example : Text summarization, translation)

- https://huggingface.co/docs/transformers/main/tasks/prompting#reasoning

- https://towardsdatascience.com/fine-tuning-large-language-models-llms-23473d763b91

#### **Prompting an LLM - Best Practices**

- Starting with a simple prompt and iterate based on the response.
- Working with recent and more capable models.
- Instructions for the model should be separated from the text these instructions apply to, using delimiters.
- Being clear, specific and descriptive about the task provided to the LLM as part of the prompt.
- For reasoning tasks, use of advanced prompting techniques such as tree of thoughts or chain-of-thought
- Test the prompt with different models to understand how consistent and robust these are.

#### **Finetuning vs Prompting**

- While prompting techniques can help achieve good response from an LLM, fine-tuning a model may work well for specific scenarios.

- Fine-tuning a smaller model, may yield a good response in a scenario where the domain is very different from what the LLM was pretrained on and prompt optimization didn't suffice.

- The need to use a smaller model due to cost, infrastructure or other limitations.

- The model needs to be trained on sensitive data and is under strict regulations in some rare scenarios.

- https://huggingface.co/docs/transformers/main/tasks/prompting#chain-of-thought

#### **Base vs Instruct models**

- Base models are good at the completing text, giving a prompt.
- For NLP taks with instructions or for conversational purpose, instruct(or chat) versions are helpful.

````python
# Instantiating the tokenizer and specifying the model
model = 'tiiuae/falcon-7b-instruct'
tokenizer = AutoTokenizer.from_pretrained(model)
````

````python
# Text generation pipeline
response = pipeline(
    'text-generation',
    model = model,
    tokenizer = tokenizer,
    torch_dtype = torch.bfloat16,
    device_map="auto"
)
````

#### **Text sentiment classification**

- Text sentiment classification assigns a label such as 'positive', 'negative' or 'neutral' to a sequence of text.
- The prompt provides the instruction and specifying the text to classify, additionally providing the beginning of the response - "Sentiment : "

````python
# Prompt instructing the model to classify the given text. The instruction is provided followed by the text for sentiment classification

prompt = """In one word, Classify the text shown in backticks into positive, negative or neutral sentiment : Please only select one among the three provided categories.
text : `This class is great to understand concepts in natural language processing.`
Sentiment : """
````

````python
# Extracting model response
sequences = response(prompt,
                    max_new_tokens = 5)
````

````python
sequences
````

````python
for sequence in sequences:
  print(f"Result: {sequence['generated_text']}")
````

#### **Named Entity Recognition**

- Named Entity Recognition is a task of extracting named entities from text, such as person, location.

````python
# Prompt instructing the model to extract named entities from the text.

Prompt = """
Provide a list of named entities from the text, shown in backticks :

text : `Claude_LLM focuses on constitutional AI, shapes LLM output guided by a set of principles`
Named Entitites :
"""
sequences = response(
    Prompt,
    max_new_tokens = 10,
    return_full_text = False,
)
````

````python
# The model has correctly identified the named entity from the given text :
for sequence in sequences:
  print(sequence['generated_text'])
````

#### **Text summarization**

- Text summarization is another generative task, where the model output relies on the input.
- While encoder-decoder models can be a good choice, decode-type transformer based models may be used.

- https://huggingface.co/docs/transformers/main/tasks/prompting#reasoning

````python
# https://kili-technology.com/large-language-models-llms/using-falcon-40b-to-build-high-quality-training-datasets
# https://www.techtarget.com/whatis/feature/12-of-the-best-large-language-models
# Prompt for text summarization :
prompt = """ Provide a summary of the text shown below in backticks

`Falcon 40B is a transformer based decoder-only model, trained on English data,
with two smaller variants :
Falcon 1B and Falcon 7B (1 Billion and 7 Billion parameters), available on SageMaker.
Falcon 40B is one of the best open-source models for natural language processing tasks such as
text generation and machine translation, able to perform accurate inference at scale`

"""

sequences = response(
    prompt,
    max_new_tokens = 45,
    do_sample = True,
    top_k = 10,
    return_full_text = False
)
````

````python
# response from the model
for sequence in sequences:
  print(f"{sequence['generated_text']}")
````

#### **QA Task**

For QA Task, the prompt is structured into logical components : context, question and the response keyword, to encourage the model to start generating an appropriate response.

````python
# Prompt for QA Task
prompt = """
Provide a response to the question using only the context below shown in three backticks :
context : ```Large language models are being increasing deployed in general problem solving across a wide range of tasks, but confined to token-level,
            left to right decision making processes, during inference. This could mean,
            they may not do well in tasks requiring strategic lookahead, exploration or where initial decisions play a pivotal role. To surmount these challenges
            the tree-of-thoughts framework has been introduced in reference :
            https://doi.org/10.48550/arXiv.2305.10601```

question : What challenges can be surmounted using the framework discussed in text provided in three backticks and what is the name of this framework?

response :
            """
sequences = response(
    prompt,
    max_new_tokens = 50,
    do_sample = True,
    top_k = 10,
    return_full_text = False
)
````

````python
# response from the model
for sequence in sequences:
  print(f"Response : {sequence['generated_text']}")
````

#### **Chain of thought Prompting**

````python
# Chain of thought prompting technique :
prompt = """

Please think to this question systematically step by step :

Question : I baked 15 cakes, gave 2 cakes to my neighbor and ate 5 cakes. My colleague
brought 6 cakes and ate 2. How many cakes remain?

Response :
I baked 15 cakes
I gave 2 cakes to my neighbor. So 15 - 2 = 13 cakes remain
I ate 5 cakes, 13 - 5 = 8 cakes remain
My colleage brought 6 cakes. So 8 + 6 = 14 cakes remain
My colleage ate 2 cakes, so 14 - 2 = 12 cakes remain.
Answer : 12

Question : There are 15 blue balls and 13 pink balls in a drawer. If all the of
pink balls and 3 blue balls were removed from the drawer for a game, how many total number of
balls remain in the drawer?

Response :
"""

sequences = response(
    prompt,
    max_new_tokens = 80,
    do_sample = True,
    return_full_text = False
)
````

````python
for sequence in sequences:
  print(f"Response: : {sequence['generated_text']}")
````
